# Object-Oriented Analysis and Design: Reservation state-machine tests

This is a toy model for learning: run it, change assumptions, and write an evidence limitation.

```python
LEGAL={('NEW','reserve'):'HELD',('HELD','confirm'):'CONFIRMED',('HELD','timeout'):'CANCELLED',('CONFIRMED','complete'):'COMPLETED'}
def transition(state,event):
    if state=='CONFIRMED' and event=='confirm': return state
    if (state,event) not in LEGAL: raise ValueError(f'illegal: {state}/{event}')
    return LEGAL[(state,event)]
assert transition('HELD','confirm')=='CONFIRMED'
assert transition('CONFIRMED','confirm')=='CONFIRMED'
try: transition('CANCELLED','confirm'); assert False
except ValueError: pass
print('State-machine constraints verified')
```
