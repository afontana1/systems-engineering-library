## Reproduce a key calculation
Alternative A scores 0.8 on cost and 0.6 on reliability, while B scores 0.5 and 0.9. At cost weight w, derive the preference switching point.
Expected reasoning: A=0.6+0.2w; B=0.9−0.4w. Equate them: 0.6w=0.3, so w=0.5. Feasibility and scaling of value scores remain separate questions.


```python
# A = .6+.2*w and B = .9-.4*w
w=(.9-.6)/(.2+.4)
print('Preference switching weight',w)
assert abs(w-.5)<1e-12
```
