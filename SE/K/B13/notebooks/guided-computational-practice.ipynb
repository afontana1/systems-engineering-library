## Reproduce a key calculation
A service backlog begins at 40 requests, with 12 new requests and 9 completions per day for 5 days. Compute the ending stock.
Expected reasoning: Backlog = 40 + 5×(12−9)=55. This is a stock balance, not a claim about service quality.


```python
backlog=40; arrival=12; completion=9; days=5
end=backlog+(arrival-completion)*days
print('New backlog',end)
assert end==55
```
