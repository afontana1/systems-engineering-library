# Introduction to Model-Based Systems Engineering: Typed traceability audit

This is a toy model for learning: run it, change assumptions, and write an evidence limitation.

```python
links=[('R-12','T-12','verify'),('R-12','SW-DISPATCH','satisfy'),('R-21','SW-VEHICLE','satisfy')]
required={'R-12','R-21'}
covered={a for a,b,k in links if k=='verify'}
print('Requirements without verification case:', sorted(required-covered))
assert required-covered == {'R-21'}
```
