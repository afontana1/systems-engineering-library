# Software Systems Engineering: Load and idempotency checks

This is a toy model for learning: run it, change assumptions, and write an evidence limitation.

```python
class Booking:
    def __init__(self): self.receipts={};self.next_id=1
    def reserve(self,key):
        if key not in self.receipts:
            self.receipts[key]=self.next_id;self.next_id+=1
        return self.receipts[key]
b=Booking()
assert b.reserve('req-42')==b.reserve('req-42')
assert len(b.receipts)==1
arrival,service=110,100
print('Backlog after 6 min (simple model):',max(0,arrival-service)*6)
assert max(0,arrival-service)*6==60
```
