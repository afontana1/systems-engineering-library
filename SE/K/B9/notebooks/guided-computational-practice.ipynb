## Reproduce a key calculation
A model relation table contains 28 requirements, of which 21 link to at least one test and 18 have both source need and test. Calculate test-link coverage and complete-chain coverage.
Expected reasoning: Test-link coverage = 21/28 = 75%; complete-chain coverage = 18/28 ≈ 64.3%. These are structural metrics, not proof that test quality or need validity is sufficient.


```python
requirements = 28; test_links = 21; complete = 18
print('Test links',test_links/requirements,'complete chain',complete/requirements)
assert (test_links/requirements)==0.75
```
