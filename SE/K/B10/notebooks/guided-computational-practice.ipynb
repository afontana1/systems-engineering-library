## Reproduce a key calculation
A sample of 25 independent results has mean 14 min and sample standard deviation 5 min. Estimate a 95% t interval for the population mean using t(24)≈2.064.
Expected reasoning: Standard error = 5/√25=1; interval ≈ 14±2.064, or [11.936,16.064] minutes, under the sampling assumptions.


```python
n=25; mean=14; s=5; tcritical=2.064
half=tcritical*s/math.sqrt(n)
print('Mean 95% t interval', (mean-half,mean+half))
assert abs(half-2.064)<1e-12
```
