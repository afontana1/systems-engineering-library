## Reproduce a key calculation
For a theoretical M/M/1 queue λ=8/hour and μ=10/hour, calculate utilization and expected time in system.
Expected reasoning: ρ=8/10=0.8; W=1/(10−8)=0.5 hours=30 minutes. M/M/1 assumptions are restrictive.


```python
lam=8.0; mu=10.0
rho=lam/mu; wait=1/(mu-lam)
print('Utilization',rho,'expected system hours',wait)
assert rho<1 and abs(wait-.5)<1e-12
```
