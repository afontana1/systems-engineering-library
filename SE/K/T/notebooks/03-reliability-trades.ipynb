# Reliability and maintenance basics

All inputs are synthetic. Execute the calculations, change one assumption, and record what conclusion changes.

```python
import math
r=math.exp(-10/500)
a=500/(500+5)
s=.95*.95
p=1-(1-.95)**2
assert abs(p-.9975)<1e-12
print('10-hour reliability:',round(r,4),'Availability:',round(a,4),'Series:',s,'Parallel:',p)
```
