# Safety constraint and stale evidence

All inputs are synthetic. Execute the calculations, change one assumption, and record what conclusion changes.

```python
def authorized(now, last_valid, zone_clear, signature_valid, revision_current):
    return (now >= last_valid and now-last_valid <= 6 and zone_clear
            and signature_valid and revision_current)
scenarios=[(12,8,True,True,True),(12,4,True,True,True),(12,8,True,False,True)]
assert [authorized(*s) for s in scenarios] == [True,False,False]
print([authorized(*s) for s in scenarios])
```
