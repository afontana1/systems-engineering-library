# Lifecycle economics and hard constraints

All inputs are synthetic. Execute the calculations, change one assumption, and record what conclusion changes.

```python
rate=.05
annuity=sum((1+rate)**(-t) for t in range(1,4))
A=1000+200*annuity
B=1200+110*annuity
choices=[('A',6,12),('B',12,24),('C',18,36)]
feasible=[name for name,service,energy in choices if service>=10 and energy<=30]
assert feasible==['B'] and B<A
print('Lifecycle costs:',round(A,2),round(B,2),'Feasible choices:',feasible)
```
