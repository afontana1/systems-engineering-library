# Factorial experiment: interactions

All inputs are synthetic. Execute the calculations, change one assumption, and record what conclusion changes.

```python
a,b,c,d=38,30,22,18
fleet=(c+d-a-b)/2
policy=(b+d-a-c)/2
interaction=(d-c)-(b-a)
assert (fleet,policy,interaction)==(-14,-6,4)
print('Fleet:',fleet,'Policy:',policy,'Interaction:',interaction)
```
