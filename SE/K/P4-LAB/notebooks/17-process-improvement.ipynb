## Right summary for skewed data

```python
from statistics import mean,median
lead=[4,5,7,9,65]
print('mean',mean(lead),'median',median(lead))
assert mean(lead)==18 and median(lead)==7
```

## A controlled paired experiment

```python
from statistics import mean,stdev
from math import sqrt
before=[20,22,18,19,27,24]
after=[18,20,16,17,24,22]
d=[a-b for a,b in zip(after,before)]
avg=mean(d);half=1.96*stdev(d)/sqrt(len(d))
print('post minus pre mean',avg,'approximate 95% normal interval',avg-half,avg+half)
assert avg<0
```
