## Requirements and exact revisions

```python
reqs={('R14','C'),('R17','C'),('R21','A')}
links={('R14','B'),('R17','B'),('R21','A')}
print('unverified current versions',sorted(reqs-links))
assert reqs-links=={('R14','C'),('R17','C')}
```
