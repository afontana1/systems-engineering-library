# Lab notebook 2 · Fault injection and service evidence
Reports are for a synthetic fixed fixture. A better conditional p90 under message loss can coexist with much worse completion.

```python
import json
from src.event_sim import Scenario,load_events,simulate
cases={s["id"]:Scenario(**s) for s in json.load(open("data/scenarios.json"))}
requests=load_events("data/requests.csv")
results={k:simulate(requests,v) for k,v in cases.items()}
for key,r in results.items(): print(key,"completed",r["completed"],"lost",r["lost_messages"],"p90",r["p90_wait_min"])
```

```python
assert results["BASE"]["completed"]==12
assert results["OUTAGE"]["completed"]==6
assert results["LOSS"]["lost_messages"]>0
assert all(r["vehicle_nonoverlap"] for r in results.values())
print("Fixture assertions satisfied, field readiness NOT established")
```
