# Lab notebook 3 · Common-random-number comparison
The confidence interval is **Monte Carlo precision**, not validation against real demand.

```python
import json
from src.event_sim import Scenario,paired_experiment
cases={s["id"]:Scenario(**s) for s in json.load(open("data/scenarios.json"))}
r=paired_experiment(cases["BASE"],cases["OUTAGE"],replications=120,seed=911)
print(r)
assert r["paired_mean_difference"]<0
```

```python
for mean_gap in [2,4,8]:
    r=paired_experiment(cases["BASE"],cases["OUTAGE"],replications=50,mean_gap=mean_gap,seed=1700)
    print("mean gap",mean_gap,"paired effect",r["paired_mean_difference"],"interval",r["approximate_normal_95pct_interval"])
```
