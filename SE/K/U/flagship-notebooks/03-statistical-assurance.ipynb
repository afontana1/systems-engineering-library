```python
from reliability_evidence import zero_failure_upper, zero_failure_sample_size, series_reliability, two_channel_redundancy, beta_posterior_mean
```

```python
print('95% upper bound, 300/0:',zero_failure_upper(300))
print('n to get p<=0.001 after zero failures:',zero_failure_sample_size(.001))
print('series R:',series_reliability(.98,.99,.97))
print('redundant with common cause:',two_channel_redundancy(.08,.02))
print('Beta(1,1) posterior mean:',beta_posterior_mean(1,1,0,300))
```
