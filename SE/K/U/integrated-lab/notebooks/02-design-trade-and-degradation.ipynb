Compare **feasibility before preference scoring**. All costs and operational inputs here are fictional.

```python
from src.northstar import Config, load_requests, dispatch
requests=load_requests("data/requests.csv")
options=[]
for n,cost in [(1,100),(2,185),(3,260)]:
    result=dispatch(requests,Config(vehicles=n))
    options.append((n,cost,result["served_fraction"],result["p90_wait_min"],result["demonstration_pass"]))
for option in options: print("vehicles,cost,served,p90,pass:",option)
```

```python
feasible=[x for x in options if x[-1]]
assert any(x[0]==2 for x in feasible)
assert not any(x[0]==1 for x in feasible)
print("feasible fleet sizes:",[x[0] for x in feasible])
```

```python
for threshold in [10,15,20,25]:
    r=dispatch(requests,Config(vehicles=2,target_p90_wait_min=threshold))
    print("p90 threshold",threshold,"fixture pass",r["demonstration_pass"])
```
