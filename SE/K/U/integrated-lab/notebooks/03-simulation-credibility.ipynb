Repeated synthetic arrival traces evaluate **model sensitivity**, not actual service reliability. The confidence interval measures Monte Carlo uncertainty only.

```python
from src.northstar import Config, synthetic_replications
model=Config()
for gap in (3.0,4.0,6.0):
    result=synthetic_replications(model,count=200,seed=20261101,mean_gap=gap)
    print(gap, result["mean_served_fraction"],result["approximate_95pct_interval_for_replication_mean"], result["fraction_of_replications_meeting_demonstration_targets"])
```

```python
short=synthetic_replications(model,count=100,seed=11)
long=synthetic_replications(model,count=400,seed=11)
w1=short["approximate_95pct_interval_for_replication_mean"][1]-short["approximate_95pct_interval_for_replication_mean"][0]
w2=long["approximate_95pct_interval_for_replication_mean"][1]-long["approximate_95pct_interval_for_replication_mean"][0]
print("CI widths: n=100",w1,"n=400",w2)
assert w2<w1
```
