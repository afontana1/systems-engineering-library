# Concept study: feasibility versus attractiveness

Review the case: eight requests per hour, six trips served per hour, three-hour event window. What cannot be inferred about waiting times from backlog alone?

*All values are synthetic teaching assumptions.*

```python
from math import ceil
arrivals=8; capacity=6; hours=3
backlog=max(0,(arrivals-capacity)*hours)
print('Illustrative backlog:',backlog,'requests')
print('Decision: capacity infeasible under sustained surge; actual wait distribution not determined.')
```
